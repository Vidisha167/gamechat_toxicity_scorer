## TRAINING DIFFERENT MODELS

### (1) IMPORTS

```python
import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor
```

### (2) LOAD DATA

```python
import scipy.sparse as sp

data1 = pd.read_csv("../data/chat/dataset_1_cleaned.csv")
X1 = sp.load_npz("../data/chat/dataset_1_features.npz")
y1 = data1["score"]

data2 = pd.read_csv("../data/chat/dataset_2_cleaned.csv")
X2 = sp.load_npz("../data/chat/dataset_2_features.npz")
y2 = data2["score"]

X1_train, X1_test, y1_train, y1_test = train_test_split(
    X1, y1, test_size=0.2, random_state=42
)
X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y2, test_size=0.2, random_state=42
)
```

### (3) TRAINING MODELS

#### Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import accuracy_score, f1_score, classification_report

# Binarize scores
y1_bin = (y1 > 0).astype(int)
y2_bin = (y2 > 0).astype(int)

X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1_bin, test_size=0.2, random_state=42)
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2_bin, test_size=0.2, random_state=42)

# Dataset 1
lr1 = LogisticRegression(max_iter=1000, n_jobs=-1)
lr1.fit(X1_train, y1_train)
preds_lr1 = lr1.predict(X1_test)
print("LOGISTIC REGRESSION - DATASET 1")
print("Accuracy:", accuracy_score(y1_test, preds_lr1))
print("F1 Score:", f1_score(y1_test, preds_lr1))
print(classification_report(y1_test, preds_lr1))
joblib.dump(lr1, "../models/dataset1_logistic_regression.pkl")

# Dataset 2
lr2 = LogisticRegression(max_iter=1000, n_jobs=-1, class_weight="balanced")
lr2.fit(X2_train, y2_train)
preds_lr2 = lr2.predict(X2_test)
print("LOGISTIC REGRESSION - DATASET 2")
print("Accuracy:", accuracy_score(y2_test, preds_lr2))
print("F1 Score:", f1_score(y2_test, preds_lr2))
print(classification_report(y2_test, preds_lr2))
joblib.dump(lr2, "../models/dataset2_logistic_regression.pkl")
```

#### Ridge Regression

```python
from sklearn.linear_model import RidgeClassifier
from sklearn.calibration import CalibratedClassifierCV

# Dataset 1
ridge1 = CalibratedClassifierCV(RidgeClassifier(), cv=3)
ridge1.fit(X1_train, y1_train)
preds_ridge1 = ridge1.predict(X1_test)
print("RIDGE CLASSIFIER - DATASET 1")
print("Accuracy:", accuracy_score(y1_test, preds_ridge1))
print("F1 Score:", f1_score(y1_test, preds_ridge1))
print(classification_report(y1_test, preds_ridge1))
joblib.dump(ridge1, "../models/dataset1_ridge_classifier.pkl")

# Dataset 2
ridge2 = CalibratedClassifierCV(RidgeClassifier(), cv=3)
ridge2.fit(X2_train, y2_train)
preds_ridge2 = ridge2.predict(X2_test)
print("RIDGE CLASSIFIER - DATASET 2")
print("Accuracy:", accuracy_score(y2_test, preds_ridge2))
print("F1 Score:", f1_score(y2_test, preds_ridge2))
print(classification_report(y2_test, preds_ridge2))
joblib.dump(ridge2, "../models/dataset2_ridge_classifier.pkl")
```

#### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

# Dataset 1
rf1 = RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=42)
rf1.fit(X1_train, y1_train)
preds_rf1 = rf1.predict(X1_test)
print("RANDOM FOREST - DATASET 1")
print("Accuracy:", accuracy_score(y1_test, preds_rf1))
print("F1 Score:", f1_score(y1_test, preds_rf1))
print(classification_report(y1_test, preds_rf1))
joblib.dump(rf1, "../models/dataset1_random_forest.pkl")

# Dataset 2
rf2 = RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=42)
rf2.fit(X2_train, y2_train)
preds_rf2 = rf2.predict(X2_test)
print("RANDOM FOREST - DATASET 2")
print("Accuracy:", accuracy_score(y2_test, preds_rf2))
print("F1 Score:", f1_score(y2_test, preds_rf2))
print(classification_report(y2_test, preds_rf2))
joblib.dump(rf2, "../models/dataset2_random_forest.pkl")
```

#### XGBoost

```python
from xgboost import XGBClassifier

# Dataset 1
xgb1 = XGBClassifier(n_estimators=100, n_jobs=-1, random_state=42, eval_metric="logloss")
xgb1.fit(X1_train, y1_train)
preds_xgb1 = xgb1.predict(X1_test)
print("XGBOOST - DATASET 1")
print("Accuracy:", accuracy_score(y1_test, preds_xgb1))
print("F1 Score:", f1_score(y1_test, preds_xgb1))
print(classification_report(y1_test, preds_xgb1))
joblib.dump(xgb1, "../models/dataset1_xgboost.pkl")

# Dataset 2
xgb2 = XGBClassifier(n_estimators=100, n_jobs=-1, random_state=42, eval_metric="logloss")
xgb2.fit(X2_train, y2_train)
preds_xgb2 = xgb2.predict(X2_test)
print("XGBOOST - DATASET 2")
print("Accuracy:", accuracy_score(y2_test, preds_xgb2))
print("F1 Score:", f1_score(y2_test, preds_xgb2))
print(classification_report(y2_test, preds_xgb2))
joblib.dump(xgb2, "../models/dataset2_xgboost.pkl")
```

### (4) COMPARING MODELS

```python
from sklearn.metrics import roc_auc_score, average_precision_score
from sklearn.metrics import (accuracy_score, f1_score, precision_score,
                              recall_score, roc_auc_score, average_precision_score)

# ============================================================
# STORE ALL RESULTS
# ============================================================

results = {
    "Dataset 1": {},
    "Dataset 2": {}
}

models_d1 = {
    "Logistic Regression": (lr1, X1_test, y1_test),
    "Ridge Classifier":    (ridge1, X1_test, y1_test),
    "Random Forest":       (rf1, X1_test, y1_test),
    "XGBoost":             (xgb1, X1_test, y1_test),
}

models_d2 = {
    "Logistic Regression": (lr2, X2_test, y2_test),
    "Ridge Classifier":    (ridge2, X2_test, y2_test),
    "Random Forest":       (rf2, X2_test, y2_test),
    "XGBoost":             (xgb2, X2_test, y2_test),
}

def get_metrics(model, X_test, y_test):
    preds = model.predict(X_test)
    proba = model.predict_proba(X_test)[:, 1]
    return {
        "Accuracy":  round(accuracy_score(y_test, preds), 4),
        "F1 Score":  round(f1_score(y_test, preds), 4),
        "Precision": round(precision_score(y_test, preds), 4),
        "Recall":    round(recall_score(y_test, preds), 4),
        "ROC-AUC":   round(roc_auc_score(y_test, proba), 4),
        "AUPRC":     round(average_precision_score(y_test, proba), 4),
    }

# ============================================================
# DATASET 1 COMPARISON
# ============================================================
print("=" * 60)
print("DATASET 1 - MODEL COMPARISON")
print("=" * 60)

for name, (model, X_test, y_test) in models_d1.items():
    results["Dataset 1"][name] = get_metrics(model, X_test, y_test)

df_results1 = pd.DataFrame(results["Dataset 1"]).T
print(df_results1)
print(f"\nBest model (F1): {df_results1['F1 Score'].idxmax()} ({df_results1['F1 Score'].max()})")
print(f"Best model (AUPRC): {df_results1['AUPRC'].idxmax()} ({df_results1['AUPRC'].max()})")

# ============================================================
# DATASET 2 COMPARISON
# ============================================================
print("\n" + "=" * 60)
print("DATASET 2 - MODEL COMPARISON")
print("=" * 60)

for name, (model, X_test, y_test) in models_d2.items():
    results["Dataset 2"][name] = get_metrics(model, X_test, y_test)

df_results2 = pd.DataFrame(results["Dataset 2"]).T
print(df_results2)
print(f"\nBest model (F1): {df_results2['F1 Score'].idxmax()} ({df_results2['F1 Score'].max()})")
print(f"Best model (AUPRC): {df_results2['AUPRC'].idxmax()} ({df_results2['AUPRC'].max()})")

# ============================================================
# SAVE BEST MODELS
# ============================================================
best1 = df_results1["F1 Score"].idxmax()
best2 = df_results2["F1 Score"].idxmax()

print("\n" + "=" * 60)
print(f"Best model for Dataset 1: {best1}")
print(f"Best model for Dataset 2: {best2}")
```

```python
def clean_text(text):
    import re
    from nltk.corpus import stopwords
    from nltk.stem import WordNetLemmatizer

    stop_words = set(stopwords.words("english"))
    lemmatizer = WordNetLemmatizer()

    text = str(text).lower()
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"[^\w\s]", "", text)
    text = re.sub(r"\d+", "", text)

    words = text.split()
    words = [w for w in words if w not in stop_words]
    words = [lemmatizer.lemmatize(w) for w in words]

    return " ".join(words)
```

### (5) EXAMPLE PREDICTION

```python
import pickle
import joblib
import scipy.sparse as sp

# Load vectorizers
with open("../models/dataset_1_tfidf.pkl", "rb") as f:
    tfidf1 = pickle.load(f)
with open("../models/dataset_2_tfidf.pkl", "rb") as f:
    tfidf2 = pickle.load(f)

# Load best models
model1 = joblib.load("../models/dataset1_xgboost.pkl")
model2 = joblib.load("../models/dataset2_xgboost.pkl")



# ============================================================
# PREDICT FUNCTION
# ============================================================

TOXIC_KEYWORDS = [
    "kys", "kill yourself", "hate", "idiot", "moron", "stupid",
    "worthless", "garbage", "trash", "noob", "rekt", "worst",
    "nobody likes you", "quit", "useless", "shut up"
]

def predict_toxicity(text, tfidf, model):
    cleaned = clean_text(text)
    features = tfidf.transform([cleaned])

    if features.nnz == 0:
        return 0.0

    proba = model.predict_proba(features)[:, 1]
    return round(float(proba[0]), 4)

def final_chat_score(text):
    score1 = predict_toxicity(text, tfidf1, model1)
    score2 = predict_toxicity(text, tfidf2, model2)

    # Keyword boost
    text_lower = text.lower()
    keyword_hit = any(kw in text_lower for kw in TOXIC_KEYWORDS)

    # Only trust score2 if very confident
    if score2 >= 0.75:
        final = round(max(score1, score2), 4)
    else:
        final = round(score1, 4)

    # Boost if keyword hit but model missed
    if keyword_hit and final < 0.5:
        final = round(max(final + 0.3, 0.5), 4)

    return score1, score2, final

# ============================================================
# TEST
# ============================================================

test_phrases = [
    # Hate speech
    "women belong in the kitchen",
    "all muslims are terrorists",
    "go back to your country",
    "black people are criminals",
    
    # Gaming toxicity
    "ur trash get rekt noob",
    "kys you absolute garbage",
    "report this idiot he is useless",
    "you are the worst player i have ever seen",
    
    # General toxic
    "i hate you so much",
    "you are so stupid and worthless",
    "nobody likes you just quit",
    "shut up you absolute moron",

    # Subtle hate speech (tricky ones)
    "women are too emotional to lead",
    "men are just better at this",
    "those people always cheat",
    "they should all go back to where they came from",

    # Obviously Clean (should be low)
    "good game guys",
    "gg well played",
    "great match everyone",
    "nice shot bro",
    "let's go team",
    "good luck have fun",
    "well done everyone great effort",
    "thanks for the game",

    # Tricky Clean (might confuse model)
    "i killed it today",
    "that was insane bro",
    "we destroyed them",
    "he is so dead",
    "this game is trash",
]

print(f"{'Phrase':<35} {'Score 1':>10} {'Score 2':>10} {'Final':>10}")
print("=" * 67)

for phrase in test_phrases:
    s1, s2, final = final_chat_score(phrase)
    print(f"{phrase:<35} {s1:>10.4f} {s2:>10.4f} {final:>10.4f}")
```
